# Step fitness function

Let $q_0$ and $q_1$ be frequencies of allele 1 at locus 0 and 1, threshold of step fitness function 0.

Let locus 0 be a mean-controlling locus and locus 0 a noise-controlling locus.


$E[z | x_0 = 0/0]$

The fitness $W_{00}$ of individuals whose noise locus is `0/0` is

```python
W_00, W_11, q_0, q_1, sigma_0, sigma_1 = sp.symbols("W_00, W_11, q_0, q_1, sigma_0, sigma_1", positive = True)
mu_0, mu_1 = sp.symbols("mu_0, mu_1", real = True)
```

```python
z, mu = sp.symbols("z, mu", real = True)
sigma = sp.Symbol("sigma", positive = True)
```

```python
sp.erf(z)
```

```python
sp.Rational(1,2)
```

```python
W = sp.Rational(1,2) * (1 - q_1)**2 * ( (1 - q_0)**2 * (1 + sp.erf(mu_0 / (sigma * sp.sqrt(2)))) \
                                        + 2 * q_0 * (1 - q_0) * (1 + sp.erf(((mu_0 + mu_1)/2) / (sigma * sp.sqrt(2)))) \
                                        + q_0**2 * (1 + sp.erf(mu_1 / (sigma * sp.sqrt(2)))) 
                                      )
```

```python
W
```

```python
W_00 = W.subs(sigma, sigma_0)
W_11 = W.subs(sigma, sigma_1)
```

```python
(W_11 - W_00).simplify()
```

```python
W_11 = (1 - q_0)**2 * (1 - q_1)**2 * sp.Rational(1,2) * (1 - sp.erf((0 - mu_0)/(sigma * sp.sqrt(2)))) \
        + 2 * q_0 * (1 - q_0) * (1 - q_1)**2 *(1 - sp.erf((0-(mu_0 + mu_1)/2)/(sigma * sp.sqrt(2))))\
        + q_0**2 * (1 - q_1)**2 * sp.Rational(1,2) * (1 - sp.erf((0 - mu_1)/(sigma * sp.sqrt(2))))
```

```python
sp.nsimplify((W_11 - W_00).subs({q_0: 0.5, q_1:0.5}))
```
