```python
out = []
for i, (m0, m1, v0, v1) in enumerate(parcombs):
    if i < 2:
        print(f"{i} {i/len(parcombs):.2f} {m0} {m1} {v0} {v1}")
        M_m = m1 - m0
        M_v = v1 - v0
        z_ref = m0.sum(axis = 1)
        c_ref = v0.sum(axis = 1).reshape(1,1)
        M = hanoi.MutEffect(mean = M_m, var = M_v, cov=np.array([[]]))
        res = hanoi.run_replicates(n_reps = 3, 
                                   max_workers = 5,
                                   n_gen = 0, 
                                   genotype = G, 
                                   mut_effect = M, 
                                   record = False, 
                                   mut_rate = 0.0, 
                                   mean_0 = z_ref,
                                   varcov_0 = c_ref, 
                                   fit_func = "fit_step",
                                   boxes = np.array([[[1], [np.inf]]])
                             )
        #out.append([[m0, m1, v0, v1, res[j][0], res[j][1].allele_freqs_next()[0,0]] for j in range(len(res))])
        out.append([[m0[0,0], m1[0,0], v0[0,1], v1[0,1], res[j][0], res[j][1].allele_freqs_next()[0,0], res[j][1].allele_freqs_next()[0,1]] for j in range(len(res))])
out = np.array(out).reshape(-1, np.array(out).shape[2])
out_df = pd.DataFrame(out, columns = ["m0", "m1", "v0", "v1", "n_gen", "allele_fixed_locus_0", "allele_fixed_locus_1"])
out_df = out_df.astype({"n_gen": "int64", "allele_fixed_locus_0": np.float64, "allele_fixed_locus_1": np.float64})


```
